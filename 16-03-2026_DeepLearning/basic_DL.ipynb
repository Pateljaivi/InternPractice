```python
import tensorflow as tf
import numpy as np
tf.random.set_seed(42)
np.random.seed(42)
```

```python
#y = 2x -1
x = np.array([-1,0,1,2,3,4],dtype=float)
y = np.array([-3,-1,1,3,5,7],dtype=float)
```

```python
model = tf.keras.Sequential([tf.keras.layers.Dense(units=1,input_shape=[1])])
#layers.Dense: A "Dense" layer means every neuron in this layer is connected to every neuron in the next.
```

```python
model.compile(optimizer='sgd',loss='mean_squared_error',metrics=['accuracy'])
```

```python
history = model.fit(x,y,epochs=500,verbose=0)
print("Final Loss:", history.history['loss'][-1]) #number to be as close to 0
print("Final Accuracy:", history.history['accuracy'][-1]) # range from 0 to 1
```

```python
weights = model.get_weights()
print(f"Learned Weight (w): {weights[0][0]}")
print(f"Learned Bias (b): {weights[1][0]}")
```

```python
prediction = model.predict(np.array([[10.0]]), verbose=0)
print("If x is 10, the model predicts y is:")
print(prediction[0][0])
```
